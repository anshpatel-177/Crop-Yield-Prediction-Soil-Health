## Model Training

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
```

creating the moddels

```python
models = {
    "Linear Regression": LinearRegression(),
    
    "Decision Tree": DecisionTreeRegressor(
        random_state=42,
        max_depth=10
    ),
    
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        random_state=42
    )
}
```

train the model

```python
trained_models = {}

for name, model in models.items():
    print(f"Training {name}...")
    
    model.fit(X_train_processed, y_train)
    
    trained_models[name] = model

print("All models trained successfully!")
```

```python
## making prediction
predictions = {}

for name, model in trained_models.items():
    predictions[name] = model.predict(X_test_processed)

print("Predictions generated successfully!")
```

```python
## Model Performane
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
results = []

for name in trained_models:
    
    y_pred = predictions[name]
    
    mae = mean_absolute_error(y_test, y_pred)
    
    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred)
    )
    
    r2 = r2_score(y_test, y_pred)
    
    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2 Score": r2
    })

results_df = pd.DataFrame(results)

results_df
```

```python
results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
)

results_df
```

Creating the soil health Analysis Function

```python
def calculate_soil_health(ph, nitrogen, phosphorus, potassium, organic_matter):

    score = 0

    # pH
    if 6.0 <= ph <= 7.5:
        score += 20
    elif 5.5 <= ph <= 8.0:
        score += 15
    else:
        score += 10

    # Nitrogen
    if nitrogen >= 80:
        score += 20
    elif nitrogen >= 50:
        score += 15
    else:
        score += 10

    # Phosphorus
    if phosphorus >= 40:
        score += 20
    elif phosphorus >= 20:
        score += 15
    else:
        score += 10

    # Potassium
    if potassium >= 150:
        score += 20
    elif potassium >= 100:
        score += 15
    else:
        score += 10

    # Organic Matter
    if organic_matter >= 2.5:
        score += 20
    elif organic_matter >= 1.5:
        score += 15
    else:
        score += 10

    # Category
    if score >= 80:
        health = "Good"
    elif score >= 60:
        health = "Moderate"
    else:
        health = "Poor"

    return score, health
```

Testing the Function

```python
score, health = calculate_soil_health(
    ph=6.8,
    nitrogen=90,
    phosphorus=45,
    potassium=170,
    organic_matter=3.0
)

print("Soil Health Score:", score)
print("Soil Health:", health)
```

Generating the Recommendations

```python
def soil_recommendations(
    ph,
    nitrogen,
    phosphorus,
    potassium,
    organic_matter
):

    recommendations = []

    if ph < 6.0:
        recommendations.append(
            "Soil is acidic. Consider suitable soil amendments."
        )

    elif ph > 7.5:
        recommendations.append(
            "Soil is alkaline. Consider suitable organic amendments."
        )

    if nitrogen < 50:
        recommendations.append(
            "Nitrogen level is low. Improve nitrogen management."
        )

    if phosphorus < 20:
        recommendations.append(
            "Phosphorus level is low. Consider phosphorus-rich fertilizer."
        )

    if potassium < 100:
        recommendations.append(
            "Potassium level is low. Consider potassium supplementation."
        )

    if organic_matter < 1.5:
        recommendations.append(
            "Organic matter is low. Add compost or organic material."
        )

    if len(recommendations) == 0:
        recommendations.append(
            "Soil parameters are within the project-defined healthy range."
        )

    return recommendations
```

## Creating the reusable prediction function

```python
def predict_crop_yield(
    year,
    district,
    crop,
    season,
    farm_area,
    soil_ph,
    nitrogen,
    phosphorus,
    potassium,
    organic_matter,
    temperature,
    rainfall,
    humidity,
    soil_moisture,
    irrigation_type,
    fertilizer_use,
    pesticide_use
):

    input_data = pd.DataFrame({
        "Year": [year],
        "District": [district],
        "Crop": [crop],
        "Season": [season],
        "Farm_Area_acres": [farm_area],
        "Soil_pH": [soil_ph],
        "Nitrogen_kg_ha": [nitrogen],
        "Phosphorus_kg_ha": [phosphorus],
        "Potassium_kg_ha": [potassium],
        "Organic_Matter_pct": [organic_matter],
        "Temperature_C": [temperature],
        "Rainfall_mm": [rainfall],
        "Humidity_pct": [humidity],
        "Soil_Moisture_pct": [soil_moisture],
        "Irrigation_Type": [irrigation_type],
        "Fertilizer_Use_kg_ha": [fertilizer_use],
        "Pesticide_Use_L_ha": [pesticide_use]
    })

    # Preprocess input
    processed_data = preprocessor.transform(input_data)

    # Predict yield
    prediction = best_model.predict(processed_data)

    return prediction[0]
```

```python
## Testing the function
yield_prediction = predict_crop_yield(
    year=2026,
    district="Lucknow",
    crop="Wheat",
    season="Rabi",
    farm_area=5,
    soil_ph=6.8,
    nitrogen=90,
    phosphorus=45,
    potassium=170,
    organic_matter=3.0,
    temperature=22,
    rainfall=550,
    humidity=65,
    soil_moisture=45,
    irrigation_type="Canal",
    fertilizer_use=120,
    pesticide_use=2
)

print(
    f"Predicted Crop Yield: {yield_prediction:.2f} tonnes/hectare"
)
```
